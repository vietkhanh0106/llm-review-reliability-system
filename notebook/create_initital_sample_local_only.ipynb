# Dataset Sampling 

This notebook creates a balanced subset of the Amazon Reviews dataset for use in the main analysis.

Due to the large size of the original dataset (~500k reviews), it is not included in the submission. Instead, sampling is performed locally on the local machine, and only the resulting sample dataset is provided. A balanced sampling strategy is applied by selecting an equal number of reviews from each rating category (1 to 5). This helps reduce class imbalance and ensures representation across different sentiment levels.


```python
import pandas as pd

# Load full dataset (local only)
df = pd.read_csv("../data/Reviews.csv")
# Sample 10,000 reviews with equal representation across scores
SAMPLE_PER_SCORE = 2000  # total = 10,000
RANDOM_STATE = 16

sampled_groups = []

for score, group in df.groupby("Score"):
    sampled_groups.append(group.sample(n=SAMPLE_PER_SCORE, random_state=RANDOM_STATE))

df_sample_10k = pd.concat(sampled_groups).reset_index(drop=True)

OUTPUT_PATH = "../data/sample_reviews_10000.csv"

df_sample_10k.to_csv(OUTPUT_PATH, index=False)

print("Saved to:", OUTPUT_PATH)
```
